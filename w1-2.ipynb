```python
products = pd.DataFrame({
    "product_id": [
        "P001", "P002", "P003", "P004",
        "P005", "P006", "P007", "P008"
    ],
    
    "product_name": [
        "Analytics Platform",
        "Cloud Storage",
        "Data Warehouse",
        "BI Dashboard",
        "AI Assistant",
        "Data Pipeline",
        "Security Suite",
        "CRM Integration"
    ],
    
    "category": [
        "Software",
        "Cloud",
        "Cloud",
        "Analytics",
        "AI",
        "Engineering",
        "Security",
        "Integration"
    ],
    
    "unit_price": [
        1200,
        800,
        2500,
        1500,
        3000,
        2200,
        1800,
        1100
    ]
})

display(products)
```

```python
orders = pd.DataFrame({
    "order_id": [
        "O001", "O002", "O003", "O004", "O005",
        "O006", "O007", "O008", "O009", "O010",
        "O011", "O012", "O013", "O014", "O015"
    ],
    
    "customer_id": [
        "C001", "C002", "C003", "C004", "C005",
        "C006", "C007", "C008", "C009", "C010",
        "C001", "C002", "C003", "C004", "C005"
    ],
    
    "product_id": [
        "P001", "P003", "P004", "P005", "P006",
        "P002", "P007", "P001", "P005", "P004",
        "P003", "P005", "P008", "P007", "P001"
    ],
    
    "order_date": [
        "2026-01-05", "2026-01-08", "2026-01-12",
        "2026-01-15", "2026-01-20", "2026-02-02",
        "2026-02-07", "2026-02-11", "2026-02-18",
        "2026-02-25", "2026-03-03", "2026-03-10",
        "2026-03-14", "2026-03-18", "2026-03-25"
    ],
    
    "quantity": [
        5, 2, 3, 4, 2,
        5, 3, 6, 2, 4,
        1, 3, 5, 2, 4
    ],
    
    "discount": [
        0.05, 0.10, 0.00, 0.05, 0.10,
        0.00, 0.05, 0.10, 0.00, 0.05,
        0.00, 0.10, 0.05, 0.00, 0.05
    ]
})

orders["order_date"] = pd.to_datetime(orders["order_date"])

display(orders)
```

```python
raw_customers = customers.copy()
raw_products = products.copy()
raw_orders = orders.copy()

print("RAW LAYER")
print("-" * 40)

print("Customers:", raw_customers.shape)
print("Products :", raw_products.shape)
print("Orders   :", raw_orders.shape)
```

```python
print("CUSTOMER MISSING VALUES")
print(raw_customers.isnull().sum())

print("\nPRODUCT MISSING VALUES")
print(raw_products.isnull().sum())

print("\nORDER MISSING VALUES")
print(raw_orders.isnull().sum())
```

```python
stg_customers = raw_customers.copy()

stg_customers["customer_name"] = (
    stg_customers["customer_name"]
    .str.strip()
)

stg_customers["segment"] = (
    stg_customers["segment"]
    .str.strip()
    .str.upper()
)

stg_customers["region"] = (
    stg_customers["region"]
    .str.strip()
    .str.upper()
)

stg_customers["country"] = (
    stg_customers["country"]
    .str.strip()
    .str.upper()
)

display(stg_customers)
```

```python
stg_products = raw_products.copy()

stg_products["product_name"] = (
    stg_products["product_name"]
    .str.strip()
)

stg_products["category"] = (
    stg_products["category"]
    .str.strip()
    .str.upper()
)

stg_products["unit_price"] = (
    stg_products["unit_price"]
    .astype(float)
)

display(stg_products)
```

```python
stg_orders = raw_orders.copy()

stg_orders["order_id"] = (
    stg_orders["order_id"]
    .str.strip()
)

stg_orders["customer_id"] = (
    stg_orders["customer_id"]
    .str.strip()
)

stg_orders["product_id"] = (
    stg_orders["product_id"]
    .str.strip()
)

stg_orders["quantity"] = (
    stg_orders["quantity"]
    .astype(int)
)

stg_orders["discount"] = (
    stg_orders["discount"]
    .astype(float)
)

display(stg_orders)
```

```python
sales_model = stg_orders.merge(
    stg_customers,
    on="customer_id",
    how="left"
)

sales_model = sales_model.merge(
    stg_products,
    on="product_id",
    how="left"
)

display(sales_model)
```

```python
sales_model["gross_sales"] = (
    sales_model["quantity"]
    * sales_model["unit_price"]
)

sales_model["discount_amount"] = (
    sales_model["gross_sales"]
    * sales_model["discount"]
)

sales_model["net_sales"] = (
    sales_model["gross_sales"]
    - sales_model["discount_amount"]
)

display(
    sales_model[
        [
            "order_id",
            "customer_name",
            "product_name",
            "quantity",
            "gross_sales",
            "discount_amount",
            "net_sales"
        ]
    ]
)
```

```python
metrics = {
    "total_revenue": sales_model["net_sales"].sum(),
    "total_orders": sales_model["order_id"].nunique(),
    "total_customers": sales_model["customer_id"].nunique(),
    "total_products": sales_model["product_id"].nunique(),
    "average_order_value": sales_model["net_sales"].mean()
}

print("METRICMIND SEMANTIC METRICS")
print("=" * 40)

for metric, value in metrics.items():
    print(f"{metric}: {value:,.2f}")
```

```python
revenue_by_region = (
    sales_model
    .groupby("region")["net_sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Revenue by Region")
print("=" * 30)

display(
    revenue_by_region
    .reset_index(name="revenue")
)
```

```python
revenue_by_category = (
    sales_model
    .groupby("category")["net_sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Revenue by Product Category")
print("=" * 35)

display(
    revenue_by_category
    .reset_index(name="revenue")
)
```

```python
top_customers = (
    sales_model
    .groupby(
        ["customer_id", "customer_name"]
    )["net_sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

top_customers.columns = [
    "customer_id",
    "customer_name",
    "revenue"
]

print("Top Customers")
print("=" * 30)

display(top_customers.head(5))
```

```python
print("METRICMIND DATA QUALITY CHECKS")
print("=" * 40)

checks = {
    "Customer IDs unique":
        stg_customers["customer_id"].is_unique,

    "Product IDs unique":
        stg_products["product_id"].is_unique,

    "Order IDs unique":
        stg_orders["order_id"].is_unique,

    "No missing customer IDs":
        stg_customers["customer_id"].notna().all(),

    "No missing product IDs":
        stg_products["product_id"].notna().all(),

    "No missing order IDs":
        stg_orders["order_id"].notna().all(),

    "No negative quantity":
        (stg_orders["quantity"] >= 0).all(),

    "No negative prices":
        (stg_products["unit_price"] >= 0).all()
}

for check, result in checks.items():
    status = "PASS" if result else "FAIL"
    print(f"{check}: {status}")
```

```python
def metricmind_response(question):

    question = question.lower()

    if "total revenue" in question or "total sales" in question:
        return (
            f"The total revenue is "
            f"${metrics['total_revenue']:,.2f}."
        )

    elif "orders" in question:
        return (
            f"There are "
            f"{metrics['total_orders']} total orders."
        )

    elif "customers" in question:
        return (
            f"There are "
            f"{metrics['total_customers']} customers."
        )

    elif "products" in question:
        return (
            f"There are "
            f"{metrics['total_products']} products."
        )

    elif "average order" in question:
        return (
            f"The average order value is "
            f"${metrics['average_order_value']:,.2f}."
        )

    elif "region" in question:
        region = revenue_by_region.idxmax()
        revenue = revenue_by_region.max()

        return (
            f"The region with the highest revenue "
            f"is {region}, with "
            f"${revenue:,.2f}."
        )

    elif "category" in question:
        category = revenue_by_category.idxmax()
        revenue = revenue_by_category.max()

        return (
            f"The highest revenue category is "
            f"{category}, with "
            f"${revenue:,.2f}."
        )

    else:
        return (
            "I can currently answer questions about "
            "revenue, orders, customers, products, "
            "regions, categories, and average order value."
        )
```

```python
def stream_response(text, delay=0.04):

    words = text.split()

    for word in words:
        print(word, end=" ", flush=True)
        time.sleep(delay)

    print()
```

```python
question = "What is the total revenue?"

print("You:", question)
print("\nMetricMind:")

response = metricmind_response(question)

stream_response(response)
```

```python
print("======================================")
print("        METRICMIND CHAT")
print("======================================")
print("Type 'exit' to stop.")
print()

while True:

    question = input("You: ")

    if question.lower() == "exit":
        print("MetricMind: Goodbye!")
        break

    response = metricmind_response(question)

    print("MetricMind:", end=" ")

    stream_response(response)
```
